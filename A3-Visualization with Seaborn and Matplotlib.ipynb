## Visualization with Seaborn & Matplotlib

### Use “Titanic” dataset from Seaborn and do the following: 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
t1= sns.load_dataset('titanic') #Loading Titanic dataset
t1.info() 
```

## Task 1: Countplot - Passenger Count by Embarked Port Create a countplot to display the number of passengers who boarded from each port (embarked), separated by sex. 

```python
t1
```

```python
sns.countplot(x='embark_town',hue='sex',data=t1,palette='pastel')
plt.title("No. of Passengers from each port in Titanic")
plt.xlabel(" Embarked Port -->")
plt.ylabel("No. of Passengers -->")
plt.show()
```

### From the above plot, it is evident that most number of passengers boarded Titanic from Southampton & the least from Queenstown

```python
color = ["Navy","Magenta"]
sns.countplot(x='embarked',hue='sex',data=t1,palette=color)
plt.title("No. of Passengers from each port in Titanic")
plt.xlabel(" Embarked -->")
plt.ylabel("No. of Passengers -->")
plt.show()
```

## 2: Swarmplot - Age Distribution across Classes Create a swarmplot to visualize the distribution of age across passenger classes (pclass), categorized by sex. 

```python
plt.figure(figsize=(12,8))
sns.swarmplot(x="class",y="age",hue="sex",data=t1,palette ="bright")
plt.title("Age Distribution across Classes, categorized by sex",fontsize=18)
plt.xlabel(" Passenger Class -->",fontsize=14)
plt.ylabel("Age -->",fontsize=14)
plt.legend(title="Sex", loc="upper right", fontsize=14)
plt.tight_layout()
plt.show()
```

### The above plot depicts that the third class was desnsly populated compared to others, also from that most passengers were between thw age group of 20-35

## Task 3: Pie Chart - Gender Distribution Create a pie chart using Matplotlib to show the percentage of male vs. female passengers. - Add percentage labels using autopct - Use explode for one slice to highlight it. 

```python
t1
```

```python
gd = t1["sex"].value_counts()
gd
```

```python
plt.pie(gd, labels=gd.index, autopct="%1.1f%%", explode=[0.1,0], startangle=0, colors= color)
plt.title("Gender Distribution (in %) chart for Titanic Passengers")
plt.show()
```

### The above Piechart is a representation of Gender of passengers onbroard Titanic which conveys that almost 2/3 of passengers were male 

## Task 4: Heatmap - Correlation Matrix Create a heatmap to show correlations between age, fare, pclass, sibsp, parch, and survived. - Use seaborn.heatmap with annotations

```python
t1
```

```python
t2 = t1.select_dtypes(include='number')
corel = t2.corr()
sns.heatmap(corel, annot=True, cmap='coolwarm')
plt.title('Titanic Correlation Heatmap')
plt.show()
```

### Darker colors indicate stronger correlations, while lighter colors indicate weaker correlations; Positive correlations (when one variable increases, the other variable tends to increase) eg- (sibsp to parch) ;  Negative correlations (when one variable increases, the other variable tends to decrease) eg- (p-class to survived)

##  Task 5: Violin Plot - Fare by Class and Gender Create a violin plot to visualize fare distribution across classes, separated by gender. 

```python
t1
```

```python
plt.figure(figsize=(16,1))
sns.violinplot(x='pclass', y='fare', hue='sex', data=t1,palette=color, width=0.8) 
plt.title("Fare by Class and Gender",fontsize=18)
plt.xlabel(" Passenger Class -->",fontsize=14)
plt.ylabel("Fare -->",fontsize=14)
plt.show()
```

### The width of the violin indicates the density of data at different values, with wider sections representing higher concentrations of data. From the above plot, the clear representation is that most number of passengers opted for 3rd class with the leaset fare & the flip side is that minimal number of passengers had paid higher price than average for the first class

```python
plt.figure(figsize=(12,8))
sns.scatterplot(x='age', y='fare', hue='survived', palette=color, data=t1)
plt.title('Scatter Plot of Age vs Fare, Colored by Survival Status')
plt.xlabel('Age --->')
plt.ylabel('Fare --->')
plt.show()
```
